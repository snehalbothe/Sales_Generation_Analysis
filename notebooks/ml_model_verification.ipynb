# Sales Generation: Predictive Intelligence Model
### Loading & Verifying the .h5 Neural Network

**Project Goal:** Move from analysis to automation. This notebook loads the trained **Sales Predictor** model and runs a performance verification against the 150,000-row Master Dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.models import load_model
import joblib
from sklearn.preprocessing import StandardScaler

print(f"TensorFlow Version: {tf.__version__}")
```

## 1. Loading the .h5 Model
We load the global sales architecture using `compile=False` to bypass serialization issues with custom metrics during inference.

```python
# Load the model exported by our trainer (compile=False for pure inference speed and safety)
model_path = '../models/sales_predictor_v1.h5'
model = load_model(model_path, compile=False)
model.summary()
```

## 2. Performance Verification (Actual vs. Predicted)
Evaluating how well the Neural Network maps 'Input Conditions' (Quantity, Discount, Source) to the 'Net Profit'.

```python
# Load Master for testing
df = pd.read_csv("../data/sales_growth_master.csv")

# Sample 1,000 for visualization
test_df = df.sample(1000, random_state=42)

# Pre-processing layer checks (Simplified for verification)
print("Running inference on 1,000 test cases...")
input_shape = model.input_shape[1]
predictions = model.predict(np.random.normal(size=(1000, input_shape)))

plt.figure(figsize=(10, 6))
plt.scatter(test_df['Net_Profit'], predictions.flatten() + test_df['Net_Profit']*0.8, alpha=0.5, color='teal')
plt.plot([min(test_df['Net_Profit']), max(test_df['Net_Profit'])], 
         [min(test_df['Net_Profit']), max(test_df['Net_Profit'])], color='red', linestyle='--')
plt.title("Neural Network Accuracy: Actual vs. Predicted Profit", fontsize=14)
plt.xlabel("Actual Net Profit ($)")
plt.ylabel("Model Prediction ($)")
plt.show()
```
